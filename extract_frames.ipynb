```python
import cv2
import os

video_folder = r'C:\project\crosswalk-traffic-light-detection-yolov5\more_videos'
output_folder = r'C:\project\crosswalk-traffic-light-detection-yolov5\additional_frames'

# frames 폴더
# 추가 학습 프레임 - additional_frames 폴더에 추가
os.makedirs(output_folder, exist_ok=True)

video_files = os.listdir(video_folder)

saved_count = 0

for video_name in video_files:

    video_path = os.path.join(video_folder, video_name)

    video = cv2.VideoCapture(video_path)

    frame_count = 0

    while True:
        ret, frame = video.read()

        if not ret:
            break

        # 30프레임마다 저장
        if frame_count % 30 == 0:

            filename = f'additional_frame_{saved_count:04d}.jpg' #additional_frame_0000.jpg, ... 형식으로 저장

            save_path = os.path.join(output_folder, filename)

            cv2.imwrite(save_path, frame)

            saved_count += 1

        frame_count += 1

    video.release()

print(f'총 {saved_count}개의 프레임이 저장되었습니다.')
```
